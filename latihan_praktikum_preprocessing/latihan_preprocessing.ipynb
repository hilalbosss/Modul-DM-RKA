```python
def get_outlier_summary(data, columns):
    results = []

    for column in columns:
        q1 = data[column].quantile(0.25)
        q3 = data[column].quantile(0.75)
        iqr = q3 - q1

        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr

        outlier_mask = (
            (data[column] < lower_bound) |
            (data[column] > upper_bound)
        )

        results.append({
            "column": column,
            "lower_bound": lower_bound,
            "upper_bound": upper_bound,
            "outlier_count": outlier_mask.sum(),
            "outlier_percentage": (
                outlier_mask.mean() * 100
            )
        })

    return pd.DataFrame(results)
```

```python
numerical_columns = x.select_dtypes(
    include=np.number
).columns.tolist()

outlier_summary = get_outlier_summary(
    x,
    numerical_columns
)

display(
    outlier_summary.sort_values(
        "outlier_count",
        ascending=False
    )
)
```

```python
def iqr_capping(data, columns):
    result = data.copy()
    bounds = {}

    for column in columns:
        q1 = result[column].quantile(0.25)
        q3 = result[column].quantile(0.75)
        iqr = q3 - q1

        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr

        result[column] = result[column].clip(
            lower=lower_bound,
            upper=upper_bound
        )

        bounds[column] = {
            "lower_bound": lower_bound,
            "upper_bound": upper_bound
        }

    return result, bounds
```

```python
CONTINUOUS_COLUMNS = [
    column for column in numerical_columns
    if column != "synthetic_id"
]

X_capped, outlier_bounds = iqr_capping(
    x,
    CONTINUOUS_COLUMNS
)
```

```python
def remove_outliers_iqr(data, columns):
    result = data.copy()

    for column in columns:
        q1 = result[column].quantile(0.25)
        q3 = result[column].quantile(0.75)
        iqr = q3 - q1

        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr

        result = result[
            result[column].between(
                lower_bound,
                upper_bound
            )
        ]

    return result.reset_index(drop=True)
```
